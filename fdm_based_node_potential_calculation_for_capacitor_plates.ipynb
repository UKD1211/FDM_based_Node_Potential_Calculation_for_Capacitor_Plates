```python
import numpy as np
import matplotlib.pyplot as plt

def calculate_inner_node_voltages(N, top_voltage):
    """
    Calculate the voltages at the inner nodes based on boundary conditions.
    Args:
        N: Number of grid points per side, including boundary.
        top_voltage: Voltage at the top boundary.
        spacing: Distance between grid points.
    Returns:
        voltages: List of voltages at the inner nodes.
    """
    # Calculate electric field E
    E = top_voltage / (N-1)

    # Define the boundary values
    boundary_top = [top_voltage] * N
    boundary_bottom = [0] * N
    boundary_left = [top_voltage - i * E for i in range(1,N-1)]
    boundary_right = [top_voltage - i * E for i in range(1,N-1)]

    # Initialize the voltages at all the nodes
    voltages = np.zeros((N, N))

    # Set boundary conditions
    voltages[0, :] = boundary_top
    voltages[-1, :] = boundary_bottom
    voltages[1:-1, 0] = boundary_left
    voltages[1:-1, -1] = boundary_right

    # Gauss-Seidel iteration to solve for the inner node voltages
    max_iterations = 10000
    tolerance = 1e-6

    for iteration in range(max_iterations):
        old_voltages = np.copy(voltages)

        # Update the voltages for each inner node
        for i in range(1, N-1):
            for j in range(1, N-1):
                voltages[i, j] = 0.25 * (voltages[i+1, j] + voltages[i-1, j] + voltages[i, j+1] + voltages[i, j-1])

        # Check for convergence

        diff = np.linalg.norm(voltages - old_voltages)
        if diff < tolerance:
            print(f"Converged after {iteration + 1} iterations.")
            break
    else:
        print("Did not converge within the maximum number of iterations.")

    # Return voltages
    return voltages

# Parameters
top_voltage = float(input("Enter the top boundary voltage (e.g., 500V): "))  # Top boundary voltage
N = (int(input("Enter the number of grid points per side, including boundary (e.g., 5): "))+1)   # Number of grid points
#spacing = float(input(f"Enter the spacing between points (d), ensure d/N is a natural number: "))  # Grid spacing

# Calculate inner node voltages
voltages = calculate_inner_node_voltages(N, top_voltage)

# Print the final inner node voltages
print("\nFinal node voltages:")
print(voltages)

# Plot the node voltages

plt.imshow(voltages, cmap='viridis', interpolation='nearest')
plt.colorbar()  # Adds a colorbar to the side.
plt.title(f"{int(top_voltage)} volt is given on the upper plate.")
plt.ylabel("Variation of voltages(volt) ->")
plt.xlabel("Ground is connected to the bottom plate.")
plt.show()
```

```python
# WITH THE Successive Over-Relaxation (SOR).

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def calculate_inner_node_voltages(N, top_voltage, omega=1.5):
    """
    Calculate the voltages at the inner nodes using SOR for faster convergence.

    Args:
        N: Number of grid points per side, including boundary.
        top_voltage: Voltage at the top boundary.
        omega: Relaxation factor (default is 1.5 for over-relaxation).

    Returns:
        voltages: Array of voltages at the inner nodes.
    """
    # Calculate electric field E
    E = top_voltage / (N - 1)

    # Define the boundary values
    boundary_top = [top_voltage] * N
    boundary_bottom = [0] * N
    boundary_left = [top_voltage - i * E for i in range(1, N - 1)]
    boundary_right = [top_voltage - i * E for i in range(1, N - 1)]

    # Initialize the voltages at all the nodes
    voltages = np.zeros((N, N))

    # Set boundary conditions
    voltages[0, :] = boundary_top
    voltages[-1, :] = boundary_bottom
    voltages[1:-1, 0] = boundary_left
    voltages[1:-1, -1] = boundary_right

    # Gauss-Seidel iteration with relaxation (SOR) to solve for the inner node voltages
    max_iterations = 10000
    tolerance = 1e-6

    for iteration in range(max_iterations):
        old_voltages = np.copy(voltages)

        # Update the voltages for each inner node using the SOR formula
        for i in range(1, N-1):
            for j in range(1, N-1):
                new_value = 0.25 * (voltages[i+1, j] + voltages[i-1, j] + voltages[i, j+1] + voltages[i, j-1])
                voltages[i, j] = (1 - omega) * old_voltages[i, j] + omega * new_value

        # Check for convergence
        diff = np.linalg.norm(voltages - old_voltages)
        if diff < tolerance:
            print(f"Converged after {iteration + 1} iterations with relaxation factor {omega}.")
            break
    else:
        print("Did not converge within the maximum number of iterations.")

    # Return voltages
    return voltages

# Parameters
top_voltage = float(input("Enter the top boundary voltage (e.g., 500V): "))  # Top boundary voltage
N = int(input("Enter the number of grid points per side, including boundary (e.g., 5): ")) + 1  # Number of grid points
omega = float(input("Enter the relaxation factor (e.g., 1.5): "))  # Relaxation factor

# Calculate inner node voltages
voltages = calculate_inner_node_voltages(N, top_voltage, omega)

# Print the final inner node voltages
print("\nFinal node voltages:")
print(voltages)

# 3D Plot: Showing Two Plates with the Grid in Between
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Create grid of x, y coordinates
x = np.linspace(0, N-1, N)
y = np.linspace(0, N-1, N)
X, Y = np.meshgrid(x, y)

# Plot the surface for the voltage distribution
surf = ax.plot_surface(X, Y, voltages, cmap='viridis', edgecolor='k')

# Add the capacitor plates
# Top plate at z = top_voltage (high voltage)
ax.plot_surface(X, Y, top_voltage * np.ones_like(voltages), color='red', alpha=0.6, label='Top Plate')

# Bottom plate at z = 0 (grounded)
ax.plot_surface(X, Y, np.zeros_like(voltages), color='blue', alpha=0.6, label='Bottom Plate')

# Labels and title
ax.set_title(f"3D Voltage Distribution Between Two Capacitor Plates ({int(top_voltage)}V Top, Grounded Bottom)")
ax.set_xlabel("X Position")
ax.set_ylabel("Y Position")
ax.set_zlabel("Voltage (V)")

# Color bar for the voltage distribution
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=5)

# Show the plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.gridspec import GridSpec
import matplotlib.patches as patches

class CoaxialCableVisualizer:
    """Visualize voltage distribution and electric field in a coaxial cable."""

    def _init_(self, V_in=100, r_inner=3, r_outer=6, num_points=100):
        """Initialize with cable parameters."""
        if r_inner <= 0 or r_outer <= 0 or r_inner >= r_outer or V_in == 0:
            raise ValueError("Invalid parameters: radii must be positive, inner < outer, V_in ≠ 0")

        self.V_in, self.r_inner, self.r_outer = V_in, r_inner, r_outer
        self.num_points = num_points
        self.ratio = self.r_outer / self.r_inner
        self.r_values = np.linspace(self.r_inner, self.r_outer, self.num_points)

    def voltage(self, r):
        """Compute voltage at radius r."""
        return self.V_in * (np.log(self.r_outer / r) / np.log(self.ratio))

    def electric_field(self, r):
        """Compute electric field strength at radius r."""
        return self.V_in / (r * np.log(self.ratio))

    def capacitance_per_length(self):
        """Calculate capacitance per unit length (F/cm)."""
        return 2 * np.pi * 8.85e-12 / np.log(self.ratio) * 1e2  # F/cm

    def generate_meshgrids(self):
        """Generate the polar and Cartesian meshgrids for visualization."""
        theta_values = np.linspace(0, 2 * np.pi, self.num_points)
        self.R, self.Theta = np.meshgrid(self.r_values, theta_values)
        self.X, self.Y = self.R * np.cos(self.Theta), self.R * np.sin(self.Theta)
        self.Z = self.voltage(self.R)

    def print_analysis(self):
        """Print analysis data."""
        max_e_field = self.electric_field(self.r_inner)
        print(f"\n--- Coaxial Cable Analysis ---")
        print(f"Maximum electric field: {max_e_field:.2f} V/cm (at inner conductor)")
        print(f"Capacitance per unit length: {self.capacitance_per_length():.2e} F/cm")
        print(f"Total voltage drop: {self.V_in:.2f} V")
        print(f"Ratio r_outer/r_inner: {self.ratio:.2f}")

    def plot(self):
        """Create visualization with all plots."""
        self.generate_meshgrids()

        # Create figure with grid layout
        fig = plt.figure(figsize=(12, 10))
        gs = GridSpec(2, 2)

        # 3D surface plot
        ax_3d = fig.add_subplot(gs[0, 0], projection='3d')
        surf = ax_3d.plot_surface(self.X, self.Y, self.Z, cmap='viridis', alpha=0.8)
        ax_3d.set_title("3D Voltage Distribution")
        ax_3d.set_xlabel("X (cm)"), ax_3d.set_ylabel("Y (cm)"), ax_3d.set_zlabel("Voltage (V)")
        fig.colorbar(surf, ax=ax_3d, shrink=0.5, label="Voltage (V)")

        # Contour plot
        ax_contour = fig.add_subplot(gs[0, 1])
        contour = ax_contour.contourf(self.X, self.Y, self.Z, levels=20, cmap='viridis')

        # Add conductors
        ax_contour.add_patch(plt.Circle((0, 0), self.r_inner, fill=False, color='red', lw=2))
        ax_contour.add_patch(plt.Circle((0, 0), self.r_outer, fill=False, color='blue', lw=2))
        ax_contour.set_aspect('equal')
        ax_contour.set_title("Voltage Contour Map")
        ax_contour.set_xlabel("X (cm)"), ax_contour.set_ylabel("Y (cm)")
        fig.colorbar(contour, ax=ax_contour, label="Voltage (V)")

        # Voltage plot
        ax_v = fig.add_subplot(gs[1, 0])
        ax_v.plot(self.r_values, self.voltage(self.r_values), 'g-', lw=2)
        ax_v.axvline(x=self.r_inner, color='r', ls='--', label=f"Inner r={self.r_inner} cm")
        ax_v.axvline(x=self.r_outer, color='b', ls='--', label=f"Outer r={self.r_outer} cm")
        ax_v.set_title("Voltage vs. Radius"), ax_v.grid(True)
        ax_v.set_xlabel("Radius (cm)"), ax_v.set_ylabel("Voltage (V)")
        ax_v.legend()

        # E-field plot
        ax_e = fig.add_subplot(gs[1, 1])
        ax_e.plot(self.r_values, self.electric_field(self.r_values), 'r-', lw=2)
        ax_e.axvline(x=self.r_inner, color='r', ls='--')
        ax_e.axvline(x=self.r_outer, color='b', ls='--')
        ax_e.set_title("Electric Field vs. Radius")
        ax_e.set_xlabel("Radius (cm)"), ax_e.set_ylabel("E-field (V/cm)")
        ax_e.grid(True)

        plt.tight_layout()
        return fig

def main():
    """Run the application with user inputs."""
    print("=== Coaxial Cable Electric Field Visualizer ===")

    try:
        # Get inputs with defaults
        get_input = lambda prompt, default: float(input(f"{prompt} [{default}]: ") or default)
        V_in = get_input("Voltage (V)", 100)
        r_inner = get_input("Inner radius (cm)", 3)
        r_outer = get_input("Outer radius (cm)", 6)

        # Create and run visualizer
        viz = CoaxialCableVisualizer(V_in, r_inner, r_outer)
        viz.print_analysis()
        fig = viz.plot()
        plt.show()

    except ValueError as e:
        print(f"Error: {e}")
    except KeyboardInterrupt:
        print("\nExiting...")
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.gridspec import GridSpec
import matplotlib.patches as patches

class CoaxialCableVisualizer:
    """Visualize voltage distribution and electric field in a coaxial cable."""

    def __init__(self, V_in=100, r_inner=3, r_outer=6, num_points=100):
        """Initialize with cable parameters."""
        if r_inner <= 0 or r_outer <= 0 or r_inner >= r_outer or V_in == 0:
            raise ValueError("Invalid parameters: radii must be positive, inner < outer, V_in ≠ 0")

        self.V_in, self.r_inner, self.r_outer = V_in, r_inner, r_outer
        self.num_points = num_points
        self.ratio = self.r_outer / self.r_inner
        self.r_values = np.linspace(self.r_inner, self.r_outer, self.num_points)

    def voltage(self, r):
        """Compute voltage at radius r."""
        return self.V_in * (np.log(self.r_outer / r) / np.log(self.ratio))

    def electric_field(self, r):
        """Compute electric field strength at radius r."""
        return self.V_in / (r * np.log(self.ratio))

    def capacitance_per_length(self):
        """Calculate capacitance per unit length (F/cm)."""
        return 2 * np.pi * 8.85e-12 / np.log(self.ratio) * 1e2  # F/cm

    def generate_meshgrids(self):
        """Generate the polar and Cartesian meshgrids for visualization."""
        theta_values = np.linspace(0, 2 * np.pi, self.num_points)
        self.R, self.Theta = np.meshgrid(self.r_values, theta_values)
        self.X, self.Y = self.R * np.cos(self.Theta), self.R * np.sin(self.Theta)
        self.Z = self.voltage(self.R)

    def print_analysis(self):
        """Print analysis data."""
        max_e_field = self.electric_field(self.r_inner)
        print(f"\n--- Coaxial Cable Analysis ---")
        print(f"Maximum electric field: {max_e_field:.2f} V/cm (at inner conductor)")
        print(f"Capacitance per unit length: {self.capacitance_per_length():.2e} F/cm")
        print(f"Total voltage drop: {self.V_in:.2f} V")
        print(f"Ratio r_outer/r_inner: {self.ratio:.2f}")

    def plot(self):
        """Create visualization with all plots."""
        self.generate_meshgrids()

        # Create figure with grid layout
        fig = plt.figure(figsize=(12, 10))
        gs = GridSpec(2, 2)

        # 3D surface plot
        ax_3d = fig.add_subplot(gs[0, 0], projection='3d')
        surf = ax_3d.plot_surface(self.X, self.Y, self.Z, cmap='viridis', alpha=0.8)
        ax_3d.set_title("3D Voltage Distribution")
        ax_3d.set_xlabel("X (cm)"), ax_3d.set_ylabel("Y (cm)"), ax_3d.set_zlabel("Voltage (V)")
        fig.colorbar(surf, ax=ax_3d, shrink=0.5, label="Voltage (V)")

        # Contour plot
        ax_contour = fig.add_subplot(gs[0, 1])
        contour = ax_contour.contourf(self.X, self.Y, self.Z, levels=20, cmap='viridis')

        # Add conductors
        ax_contour.add_patch(plt.Circle((0, 0), self.r_inner, fill=False, color='red', lw=2))
        ax_contour.add_patch(plt.Circle((0, 0), self.r_outer, fill=False, color='blue', lw=2))
        ax_contour.set_aspect('equal')
        ax_contour.set_title("Voltage Contour Map")
        ax_contour.set_xlabel("X (cm)"), ax_contour.set_ylabel("Y (cm)")
        fig.colorbar(contour, ax=ax_contour, label="Voltage (V)")

        # Voltage plot
        ax_v = fig.add_subplot(gs[1, 0])
        ax_v.plot(self.r_values, self.voltage(self.r_values), 'g-', lw=2)
        ax_v.axvline(x=self.r_inner, color='r', ls='--', label=f"Inner r={self.r_inner} cm")
        ax_v.axvline(x=self.r_outer, color='b', ls='--', label=f"Outer r={self.r_outer} cm")
        ax_v.set_title("Voltage vs. Radius"), ax_v.grid(True)
        ax_v.set_xlabel("Radius (cm)"), ax_v.set_ylabel("Voltage (V)")
        ax_v.legend()

        # E-field plot
        ax_e = fig.add_subplot(gs[1, 1])
        ax_e.plot(self.r_values, self.electric_field(self.r_values), 'r-', lw=2)
        ax_e.axvline(x=self.r_inner, color='r', ls='--')
        ax_e.axvline(x=self.r_outer, color='b', ls='--')
        ax_e.set_title("Electric Field vs. Radius")
        ax_e.set_xlabel("Radius (cm)"), ax_e.set_ylabel("E-field (V/cm)")
        ax_e.grid(True)

        plt.tight_layout()
        return fig

def main():
    """Run the application with user inputs."""
    print("=== Coaxial Cable Electric Field Visualizer ===")

    try:
        # Get inputs with defaults
        get_input = lambda prompt, default: float(input(f"{prompt} [{default}]: ") or default)
        V_in = get_input("Voltage (V)", 100)
        r_inner = get_input("Inner radius (cm)", 3)
        r_outer = get_input("Outer radius (cm)", 6)

        # Create and run visualizer
        viz = CoaxialCableVisualizer(V_in, r_inner, r_outer)
        viz.print_analysis()
        fig = viz.plot()
        plt.show()

    except ValueError as e:
        print(f"Error: {e}")
    except KeyboardInterrupt:
        print("\nExiting...")

if __name__ == "__main__":
    main()
```

```python
import numpy as np
import matplotlib.pyplot as plt

class CoaxialCableFDM:
    """Finite Difference Method (FDM) for solving voltage distribution in a coaxial cable."""

    def __init__(self, V_in=100, r_inner=3, r_outer=6, num_points=100, tolerance=1e-5, max_iterations=10000):
        """Initialize parameters."""
        if r_inner <= 0 or r_outer <= 0 or r_inner >= r_outer or V_in == 0:
            raise ValueError("Invalid parameters: radii must be positive, inner < outer, V_in ≠ 0")

        self.V_in = V_in
        self.r_inner = r_inner
        self.r_outer = r_outer
        self.num_points = num_points
        self.tolerance = tolerance
        self.max_iterations = max_iterations

        # Define the discretized grid
        self.r_values = np.linspace(self.r_inner, self.r_outer, num_points)
        self.V = np.zeros(num_points)  # Potential array
        self.V[0] = V_in  # Inner conductor voltage
        self.V[-1] = 0  # Outer conductor grounded

    def solve_fdm(self):
        """Solve Laplace's equation using the Finite Difference Method."""
        iterations = 0
        error = 1.0

        while error > self.tolerance and iterations < self.max_iterations:
            old_V = self.V.copy()

            # Apply the finite difference update rule (Gauss-Seidel method)
            for i in range(1, self.num_points - 1):
                self.V[i] = 0.5 * (old_V[i - 1] + old_V[i + 1])

            error = np.max(np.abs(self.V - old_V))
            iterations += 1

        print(f"FDM converged in {iterations} iterations with error {error:.2e}")
        return iterations

    def plot_voltage_distribution(self):
        """Plot the voltage distribution inside the coaxial cable."""
        plt.figure(figsize=(8, 5))
        plt.plot(self.r_values, self.V, 'g-', lw=2)
        plt.axvline(x=self.r_inner, color='r', ls='--', label=f"Inner r={self.r_inner} cm")
        plt.axvline(x=self.r_outer, color='b', ls='--', label=f"Outer r={self.r_outer} cm")
        plt.title("Voltage Distribution using FDM")
        plt.xlabel("Radius (cm)")
        plt.ylabel("Voltage (V)")
        plt.legend()
        plt.grid()
        plt.show()

if __name__ == "__main__":
    # Initialize the solver
    fdm_solver = CoaxialCableFDM()

    # Solve using FDM
    iterations = fdm_solver.solve_fdm()

    # Plot results
    fdm_solver.plot_voltage_distribution()
```

```python
import numpy as np

class CoaxialCableFDM:
    """Solve voltage distribution using the Finite Difference Method (FDM) for a coaxial cable."""

    def __init__(self, V_in=100, r_inner=3, r_outer=6, num_points=100, tol=1e-6):
        """Initialize with cable parameters."""
        if r_inner <= 0 or r_outer <= 0 or r_inner >= r_outer or V_in == 0:
            raise ValueError("Invalid parameters: radii must be positive, inner < outer, V_in ≠ 0")

        self.V_in, self.r_inner, self.r_outer = V_in, r_inner, r_outer
        self.num_points = num_points
        self.tol = tol
        self.dr = (r_outer - r_inner) / (num_points - 1)
        self.r_values = np.linspace(self.r_inner, self.r_outer, self.num_points)
        self.voltage = np.zeros_like(self.r_values)
        self.voltage[0] = V_in  # Boundary condition at inner conductor
        self.iterations = 0

    def solve(self):
        """Solve for voltage distribution using FDM and Gauss-Seidel iteration."""
        max_diff = float('inf')
        while max_diff > self.tol:
            max_diff = 0
            for i in range(1, self.num_points - 1):
                new_v = 0.5 * (self.voltage[i - 1] + self.voltage[i + 1])
                diff = abs(new_v - self.voltage[i])
                self.voltage[i] = new_v
                max_diff = max(max_diff, diff)
            self.iterations += 1

    def get_results(self):
        """Return the number of iterations and voltage distribution."""
        return self.iterations, self.voltage

def main():
    """Run the FDM solver with user inputs."""
    print("=== Coaxial Cable Voltage Solver (FDM) ===")

    try:
        # Get inputs with defaults
        get_input = lambda prompt, default: float(input(f"{prompt} [{default}]: ") or default)
        V_in = get_input("Voltage (V)", 100)
        r_inner = get_input("Inner radius (cm)", 3)
        r_outer = get_input("Outer radius (cm)", 6)

        # Create and run solver
        solver = CoaxialCableFDM(V_in, r_inner, r_outer)
        solver.solve()
        iterations, _ = solver.get_results()

        print(f"Number of iterations: {iterations}")

    except ValueError as e:
        print(f"Error: {e}")
    except KeyboardInterrupt:
        print("\nExiting...")

if __name__ == "__main__":
    main()
```

```python
import numpy as np
import matplotlib.pyplot as plt

class CoaxialCableFDM:
    def __init__(self, V_in=100, r_inner=3, r_outer=6, dr=0.1, tol=1e-5, max_iter=10000):
        self.V_in = V_in
        self.r_inner = r_inner
        self.r_outer = r_outer
        self.dr = dr  # Radial step size
        self.tol = tol  # Convergence tolerance
        self.max_iter = max_iter
        self.r_values = np.arange(self.r_inner, self.r_outer + self.dr, self.dr)
        self.num_points = len(self.r_values)
        self.voltage_fdm = np.zeros(self.num_points)
        self.voltage_fdm[0] = V_in  # Inner conductor potential

    def analytical_voltage(self, r):
        """Analytical solution for coaxial cable voltage."""
        return self.V_in * np.log(self.r_outer / r) / np.log(self.r_outer / self.r_inner)

    def solve_fdm(self):
        """Solving the Laplace equation in cylindrical coordinates using FDM."""
        iterations = 0
        V_old = np.copy(self.voltage_fdm)
        error = float('inf')

        while error > self.tol and iterations < self.max_iter:
            for i in range(1, self.num_points - 1):
                r = self.r_values[i]
                self.voltage_fdm[i] = (V_old[i-1] * (r - self.dr) + V_old[i+1] * (r + self.dr)) / (2 * r)

            error = np.max(np.abs(self.voltage_fdm - V_old))
            V_old = np.copy(self.voltage_fdm)
            iterations += 1

        print(f"FDM Converged in {iterations} iterations with error: {error:.2e}")
        return iterations

    def percentage_error(self):
        """Calculate the error between Analytical and FDM results."""
        analytical = self.analytical_voltage(self.r_values)
        error = np.abs((self.voltage_fdm - analytical) / analytical) * 100
        return error

    def plot_voltage_vs_radius(self):
        """Plot voltage vs. radial distance."""
        analytical = self.analytical_voltage(self.r_values)
        plt.figure(figsize=(8, 6))
        plt.plot(self.r_values, self.voltage_fdm, 'r-', label='FDM Solution')
        plt.plot(self.r_values, analytical, 'b--', label='Analytical Solution')
        plt.xlabel("Radial Distance (cm)")
        plt.ylabel("Voltage (V)")
        plt.title("Voltage vs. Radial Distance")
        plt.legend()
        plt.grid()
        plt.show()

    def plot_equipotential_and_radial_lines(self):
        """Show Equipotential and Radial Lines Perpendicular."""
        r_mesh, theta_mesh = np.meshgrid(self.r_values, np.linspace(0, 2 * np.pi, 100))
        x, y = r_mesh * np.cos(theta_mesh), r_mesh * np.sin(theta_mesh)
        voltage_mesh = np.tile(self.voltage_fdm, (100, 1))

        plt.figure(figsize=(8, 6))
        plt.contour(x, y, voltage_mesh, levels=15, cmap='viridis')
        plt.quiver(x, y, np.cos(theta_mesh), np.sin(theta_mesh), color='r', scale=20)
        plt.xlabel("X (cm)")
        plt.ylabel("Y (cm)")
        plt.title("Equipotential & Radial Lines Perpendicular")
        plt.axis('equal')
        plt.show()

    def run(self):
        """Run the full analysis."""
        iterations = self.solve_fdm()
        error = self.percentage_error()
        print(f"Average Percentage Error: {np.mean(error):.2f}%")
        self.plot_voltage_vs_radius()
        self.plot_equipotential_and_radial_lines()

if __name__ == "__main__":
    cable_sim = CoaxialCableFDM(V_in=100, r_inner=3, r_outer=6, dr=0.1)
    cable_sim.run()

```
